Question 1: What is hypothesis testing in statistics?

Ans:

Hypothesis testing is a statistical method used to make decisions about a population based on sample data.

It helps determine whether there is enough evidence to support or reject a claim (hypothesis).

The process begins by stating two hypotheses: the null hypothesis (H₀) and the alternative hypothesis (H₁).

A suitable statistical test is then performed, and the p-value is compared with the chosen significance level (α).

If the p-value is less than the significance level, the null hypothesis is rejected.

If the p-value is greater than or equal to the significance level, the null hypothesis is not rejected.

Hypothesis testing is widely used in research, business, medicine, and science to make reliable decisions based on data.

Question 2: What is the null hypothesis, and how does it differ from the alternative hypothesis?

Ans:

The null hypothesis (H₀) is a statement that there is no effect, no difference, or no relationship between variables. It assumes that any observed difference is due to chance.
The alternative hypothesis (H₁ or Ha) is a statement that there is an effect, difference, or relationship between variables. It is accepted when there is enough evidence to reject the null hypothesis.

###Difference between Null Hypothesis and Alternative Hypothesis:

•Null Hypothesis (H₀): There is no effect or difference.

•Alternative Hypothesis (H₁): There is an effect or difference.

•Null Hypothesis (H₀): It is the default assumption.

•Alternative Hypothesis (H₁): It is the research claim.

•Null Hypothesis (H₀): It assumes results occur due to chance.

•Alternative Hypothesis (H₁): It assumes results are not due to chance.

•Null Hypothesis (H₀): It is rejected if sufficient evidence is found.

•Alternative Hypothesis (H₁): It is accepted when the null hypothesis is rejected.

•Null Hypothesis (H₀): It is represented by H₀.

•Alternative Hypothesis (H₁): It is represented by H₁ or Ha.

Question 3: Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.

Ans: The significance level (α) is the probability of rejecting the null hypothesis (H₀) when it is actually true. It is chosen before conducting the test, with common values being 0.05 (5%) and 0.01 (1%).

###The significance level helps determine the outcome of a hypothesis test:

•If the p-value is less than or equal to α, reject the null hypothesis (H₀).

•If the p-value is greater than α, fail to reject the null hypothesis (H₀).

Thus, the significance level is used to decide whether the test results are statistically significant or are likely due to chance.

Question 4: What are Type I and Type II errors? Give examples of each.

Ans:

In hypothesis testing, Type I and Type II errors are two kinds of decision errors.

###Type I Error (α):
Rejecting the null hypothesis (H₀) when it is actually true. This is called a false positive.

###Example:
A medical test shows that a healthy person has a disease.

###Type II Error (β):
Failing to reject the null hypothesis (H₀) when it is actually false. This is called a false negative.

###Example:
A medical test shows that a person with a disease is healthy.

Thus, Type I error means detecting an effect that does not exist, while Type II error means failing to detect an effect that actually exists.

Question 5: What is the difference between a Z-test and a T-test? Explain when to use each.

Answer:

A Z-test and a T-test are statistical tests used to compare sample means, but they are used in different situations.

Z-test: Used when the population standard deviation is known and the sample size is large (usually n ≥ 30).

T-test: Used when the population standard deviation is unknown and the sample size is small (usually n < 30).

###Difference:

###Z-test:
Population standard deviation known; large sample size; uses the Z-distribution.

###T-test:
Population standard deviation unknown; small sample size; uses the t-distribution.

Thus, use a Z-test for large samples with known population variance, and a T-test for small samples when the population variance is unknown.

Question 6: Write a Python program to generate a binomial distribution with n = 10 and p = 0.5, then plot its histogram.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
n = 10       # Number of trials
p = 0.5      # Probability of success
size = 1000  # Number of samples

# Generate binomial distribution data
data = np.random.binomial(n, p, size)

# Display first 10 values
print("Sample Data:", data[:10])

# Plot histogram
plt.hist(data, bins=range(0, n+2), edgecolor='black')
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")
plt.show()
```

Question 7: Implement hypothesis testing using Z-statistics for a sample dataset in Python. Show the Python code and interpret the results.

```python
import numpy as np
from scipy.stats import norm

# Sample data
sample_data = [
    49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
    50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
    50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
    50.3, 50.4, 50.0, 49.7, 50.5, 49.9
]

# Hypothesized population mean
mu = 50

# Sample statistics
mean = np.mean(sample_data)
std = np.std(sample_data, ddof=1)
n = len(sample_data)

# Z-statistic
z = (mean - mu) / (std / np.sqrt(n))

# Two-tailed p-value
p_value = 2 * (1 - norm.cdf(abs(z)))

print("Sample Mean:", round(mean, 3))
print("Z-Statistic:", round(z, 3))
print("P-Value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject the Null Hypothesis (H0)")
else:
    print("Fail to Reject the Null Hypothesis (H0)")
```

Question 8: Write a Python script to simulate data from a normal distribution and calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Generate random data from a normal distribution
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=100)

# Calculate sample statistics
mean = np.mean(data)
std = np.std(data, ddof=1)
n = len(data)

# Calculate 95% confidence interval
confidence = 0.95
t_value = stats.t.ppf((1 + confidence) / 2, df=n-1)
margin_error = t_value * (std / np.sqrt(n))

lower = mean - margin_error
upper = mean + margin_error

# Print results
print("Sample Mean:", round(mean, 2))
print("95% Confidence Interval: ({:.2f}, {:.2f})".format(lower, upper))

# Plot histogram
plt.hist(data, bins=10, edgecolor='black')
plt.axvline(mean, color='red', linestyle='--', label='Mean')
plt.title("Normal Distribution Data")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

Question 9: Write a Python function to calculate the Z-scores from a dataset and visualize the standardized data using a histogram. Explain what the Z-scores represent in terms of standard deviations from the mean.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore

# Sample dataset
data = np.array([45, 50, 52, 48, 47, 51, 49, 53, 46, 54])

# Function to calculate Z-scores
def calculate_z_scores(data):
    return zscore(data)

# Calculate Z-scores
z_scores = calculate_z_scores(data)

# Print Z-scores
print("Original Data:", data)
print("Z-Scores:", np.round(z_scores, 2))

# Plot histogram of standardized data
plt.hist(z_scores, bins=5, edgecolor='black')
plt.title("Histogram of Z-Scores")
plt.xlabel("Z-Score")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

###Explanation:

A Z-score indicates how many standard deviations a data value is from the mean.

Z = 0 → The value is equal to the mean.

Z > 0 → The value is above the mean.

Z < 0 → The value is below the mean.

|Z| > 2 → The value may be considered unusual or an outlier.

The histogram shows the distribution of the standardized (Z-score) data around the mean of 0.
