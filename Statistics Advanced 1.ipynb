Question 1: What is a random variable in probability theory?

Ans: A random variable is a variable that assigns a numerical value to each possible outcome of a random experiment. It is usually denoted by symbols such as X or Y. The value of a random variable depends on the outcome of the experiment and is used to represent uncertain events mathematically. Examples include the number obtained when rolling a die or the number of heads in coin tosses.

Question 2: What are the types of random variables?

Ans: There are two main types of random variables:

#1.Discrete Random Variable
A discrete random variable takes countable values.
It represents outcomes that can be counted.

Example: Number of students in a class, number of heads obtained when tossing a coin three times.

#2.Continuous Random Variable
A continuous random variable takes any value within a given range.
It represents measurements and can have infinitely many values.

Example: Height of a person, weight of an object, temperature, or time taken to complete a task.

Question 4: What is a binomial distribution, and how is it used in probability?

Ans: A binomial distribution is a probability distribution that shows the probability of getting a fixed number of successes in a fixed number of independent trials.

#Characteristics:

There are a fixed number of trials.

Each trial has only two possible outcomes: Success or Failure.

The probability of success remains the same for every trial.

Each trial is independent of the others.

#Uses in Probability:

It is used to calculate the probability of a certain number of successes.

It is commonly used in coin tosses, quality control, surveys, and medical testing.

It helps in predicting outcomes when there are only two possible results.

Question 6: What is the Central Limit Theorem (CLT), and why is it critical in statistics?

Ans: The Central Limit Theorem (CLT) states that the sampling distribution of the sample mean becomes approximately normal as the sample size increases, even if the original population is not normally distributed.

#Characteristics:

It applies to large sample sizes (usually n ≥ 30).

The sample mean follows a normal distribution.

It works for both normal and non-normal populations.

It is one of the most important theorems in statistics.

#Importance:

It helps in making statistical inferences.

It is used to calculate confidence intervals.

It supports hypothesis testing.
It allows researchers to estimate population parameters from sample data.

It is widely used in data analysis, research, economics, and machine learning.

Question 9: Write a Python program to generate 1000 random numbers from a normal distribution with mean = 50 and standard deviation = 5. Compute its mean and standard deviation using NumPy, and draw a histogram to visualize the distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 random numbers
data = np.random.normal(loc=50, scale=5, size=1000)

# Compute mean and standard deviation
mean = np.mean(data)
std = np.std(data)

print("Mean =", mean)
print("Standard Deviation =", std)

# Draw histogram
plt.hist(data, bins=30, edgecolor='black')
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

Question 10: You are working as a data analyst for a retail company. Explain how you would apply the Central Limit Theorem (CLT) to estimate the average sales with a 95% confidence interval. Write the Python code to compute the mean sales and its confidence interval.

Ans: The Central Limit Theorem (CLT) states that the sampling distribution of the sample mean is approximately normal for a sufficiently large sample size. Using the given sales data, we first calculate the mean and standard deviation. Then, we compute the standard error and use a 95% confidence interval to estimate the true average daily sales of the company.

```python
import numpy as np

# Daily sales data
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

# Calculate mean
mean = np.mean(daily_sales)

# Calculate sample standard deviation
std = np.std(daily_sales, ddof=1)

# Number of observations
n = len(daily_sales)

# Standard Error
se = std / np.sqrt(n)

# 95% Confidence Interval
lower = mean - 1.96 * se
upper = mean + 1.96 * se

print("Mean Sales =", mean)
print("95% Confidence Interval = ({:.2f}, {:.2f})".format(lower, upper))
```
